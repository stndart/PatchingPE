```python
from pathlib import Path
import pefile
```

```python
orig_base = Path('../fake_neomon_host/shifts/orig/')
orig_p = orig_base / '10800000.dll'
```

```python
bases = [f'1{hex(i)[2:]}{hex(i)[2:]}{hex(15-i)[2:]}0000' for i in range(16)]
```

```python
pe = pefile.PE(str(orig_p))

for b in bases:
    ibase = int(b, 16)
    pe.OPTIONAL_HEADER.ImageBase = ibase

    fn = orig_base / (b + '.dll')
    pe.write(fn)

pe.close()
```
